```python
#W6A16
n, m = map(int,input("Nhập số hàng n và số cột m (cách nhau bởi dấu cách):").split())
matrix = []
print(f"nhập {n} hàng, mỗi hàng gồm {m} số nguyên:")
for i in range(n):
    row = list(map(int, input(f"hàng{i +1}").split()))
    matrix.append(row)
print("Ma trận đã nhập")
for row in matrix:
    print(' '.join(f"{num:>4}" for num in row))
```

```python
#W6A17
n = int(input("nhập kích thước ma trận vuông n: "))
matrix = [] 
for i in range(n):
    row = list(map(int,input(f"hàng{i +1}: ").split()))
    matrix.append(row)
duong_cheo_chinh = [matrix[i][i] for i in range(n)]
duong_cheo_phu = [matrix[i][n-1-i] for i in range(n)]
print('đường chéo chính:', duong_cheo_chinh)
print('duong_cheo_phu', duong_cheo_phu)
```

```python
#W6A18
n, m, k = map(int, input("nhập n, m, k cách nhau bởi dấu cách:").split())
matrix = []
print(f"Nhập ma trận {n} hàng, mỗi hàng {m} số nguyên:")
for i in range(n):
    row = list(map(int, input(f"Hàng {i+1}: ").split()))
    matrix.append(row)
if 0 <= k < m:
    col_sum = sum(matrix[i][k] for i in range(n))
    print(f"tổng các phần tử ở cột {k} là:", col_sum)
elif 0 <= k < n:
    row_sum = sum(matrix[k])
    print(f"tổng các phần tử ở hàng {k} là:", row_sum)
else:
    print("Giá trị k không hợp lệ!")
```

```python
#W6A24
def move_zeros(arr, to_front=True):
    zeros = [x for x in arr if x == 0]
    non_zeros = [x for x in arr if x != 0]
    return zeros + non_zeros if to_front else non_zeros + zeros

# Ví dụ
arr = [0, 2, 3, 2, 4, 0, 0, 0, 9]
print("Về đầu:", move_zeros(arr, to_front=True))
print("Về cuối:", move_zeros(arr, to_front=False))

```

```python
#W6A21
words = input("Nhập chuỗi: ").split()   # tách thành list các từ
mapping = {}

for i, word in enumerate(words):        # duyệt từng từ với chỉ số i
    if word not in mapping:
        mapping[word] = []              # nếu chưa có thì tạo list rỗng
    mapping[word].append(i)             # thêm vị trí i vào list

# chuyển list thành tuple
result = {k: tuple(v) for k, v in mapping.items()}

print(result)
```

```python
#W6A26
def local_max_sum(arr):
    total = 0
    for i in range(1, len(arr) - 1):
        if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
            total += arr[i]
    return total
arr = list(map(int,input().split()))
print(local_max_sum(arr))
```

```python
#W6A27
def vi_tri_nho_hon_x(arr, x):
    positions = [i for i, val in enumerate(arr) if val < x]
    if positions:
        print("Vị trí các số nhỏ hơn", x, "là:", positions)
    else:
        print(-1)

# Ví dụ sử dụng
arr = [5, 2, 8, 1, 9]
x = 6
vi_tri_nho_hon_x(arr, x)
```

```python
#W6A28
def kiem_tra_day(arr):
    tang = all(arr[i] < arr[i+1] for i in range(len(arr)-1))
    giam = all(arr[i] > arr[i+1] for i in range(len(arr)-1))
    if tang:
        print('tang')
    elif giam:
        print('giam')
    else:
        print('vo danh')
arr = list(map(int,input().split()))
print(kiem_tra_day(arr))
```
