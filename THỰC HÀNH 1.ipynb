```python
a, b = map(int, input("Nhập hai số nguyên a và b (cách nhau bằng dấu cách): ").split())
tong = a + b
hieu = a - b
tich = a * b 
phan_nguyen = a // b  
phan_du = a % b
ket_qua_chia_thuc = a / b 
print(f"Tong: {tong}")
print(f"Hieu: {hieu}")
print(f"Tich: {tich}")
print(f"Phan nguyen: {phan_nguyen}")
print(f"Phan du: {phan_du}")
print(f"Ket qua chia thuc: {ket_qua_chia_thuc:.2f}")
```

```python
du_lieu_diem = input("Nhập 6 đầu điểm (a1 b1 c1 a2 b2 a3), cách nhau bằng dấu cách: ")
a1, b1, c1, a2, b2, a3 = map(float, du_lieu_diem.split())
tong_diem_he_so = (a1 + b1 + c1) + (a2 + b2) * 2 + a3 * 3
diem_trung_binh = tong_diem_he_so / 10
print(f"Điểm trung bình môn học là: {diem_trung_binh:.1f}")
```

```python
gia_goc_usd = float(input("Nhập giá đồng hồ (USD): "))
PHI_VAN_CHUYEN = 10.0
THUE_XUAT_NHAP_CANH_RATE = 0.30  
THUE_VAT_RATE = 0.10             
gia_truoc_thue = gia_goc_usd + PHI_VAN_CHUYEN
tien_thue_xuat_nhap_canh = gia_truoc_thue * THUE_XUAT_NHAP_CANH_RATE
co_so_tinh_vat = gia_truoc_thue + tien_thue_xuat_nhap_canh
tien_thue_vat = co_so_tinh_vat * THUE_VAT_RATE
tong_tien_phai_tra = co_so_tinh_vat + tien_thue_vat
print(f"Tổng số tiền Đạt phải trả là: {tong_tien_phai_tra:.2f} USD")
```

```python
n_str = input("Nhập vào một số nguyên dương n: ")
n = int(n_str)
if n <= 0:
   print("Lỗi: n phải là một số nguyên dương (n > 0).")
else:
   so_sao_thu_n = 3 * (n ** 2) - 3 * n + 1
   print(f"Số sao thứ {n} là: {so_sao_thu_n}")
```

```python
ky_tu = input("Mời nhập vào một kí tự: ")
if ky_tu.isupper():
    print(f"Chữ thường tương ứng là: {ky_tu.lower()}")
elif ky_tu.islower():
    print(f"Chữ hoa tương ứng là: {ky_tu.upper()}")
else:
    print("Kí tự nhập vào không phải là chữ cái.")
```
