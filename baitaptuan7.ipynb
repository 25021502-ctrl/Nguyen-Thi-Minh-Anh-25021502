```python
#W7A17
def majority_element(arr):
    candidate = None
    count = 0
    for num in arr:
        if count == 0:
            candidate = num 
        count += (1 if num == candidate else -1)
    if arr.count(candidate) > len(arr) // 2:
        return candidate
    return None
arr = list(map(int,input().split()))
print(majority_element(arr))


```

```python
#W7A18
def conditional_sort(arr):
    even = sorted([x for x in arr if x % 2 == 0])
    odd = sorted([x for x in arr if x % 2 == 1], reverse = True)
    return even + odd
arr = list(map(int,input().split()))
print(conditional_sort(arr))
```

```python
#W7A23
def dem_ten(xau):
    danh_sach_tu = xau.split()
    so_luong = 0
    for tu in danh_sach_tu:
        if len(tu) >= 2:
            la_chu = True
            for ky_tu in tu:
                if not ('A' <= ky_tu <= 'Z' or 'a' <= ky_tu <= 'z'):
                    la_chu = False
                    break
            if la_chu and 'A' <= tu[0] <= 'Z':
                hop_le = True
                for ky_tu in tu[1:]:
                    if not ('a' <= ky_tu <= 'z'):
                        hop_le = False
                        break
                if hop_le:
                    so_luong += 1
    return so_luong

# Ví dụ
xau = "Nguyen van A la ban cua Minh Anh"
print(dem_ten(xau))  # Kết quả: 3
```

```python
#W7A24
def xoa_tu_lien_ke_trung(xau):
    danh_sach_tu = xau.split()
    ket_qua = []
    if len(danh_sach_tu) > 0:
        ket_qua.append(danh_sach_tu[0])
        vi_tri = 1
        while vi_tri < len(danh_sach_tu):
            if danh_sach_tu[vi_tri] != danh_sach_tu[vi_tri - 1]:
                ket_qua.append(danh_sach_tu[vi_tri])
            vi_tri += 1
    return ' '.join(ket_qua)

# Ví dụ
xau = "hello hello world world world python python"
print(xoa_tu_lien_ke_trung(xau))  # Kết quả: "hello world python"
```

```python
#W7A25
def tim_so_lon_nhat(xau_so, k):
    ngan_xep = []
    so_luong_bo = len(xau_so) - k
    vi_tri = 0
    while vi_tri < len(xau_so):
        chu_so_hien_tai = xau_so[vi_tri]
        while len(ngan_xep) > 0 and so_luong_bo > 0 and ngan_xep[-1] < chu_so_hien_tai:
            ngan_xep.pop()
            so_luong_bo -= 1
        ngan_xep.append(chu_so_hien_tai)
        vi_tri += 1
    ket_qua = ''
    vi_tri = 0
    while vi_tri < k:
        ket_qua += ngan_xep[vi_tri]
        vi_tri += 1
    return ket_qua

# Ví dụ
xau_so = "192837465"
k = 5
print(tim_so_lon_nhat(xau_so, k))  # Kết quả: "98465"
```

```python
#W7A27
def is_valid_brackets(s):
    stack = []
    pairs = { ')':'(', '}':'{', ']':'[', '>':'<'}
    for char in s:
        if char in pairs.value():
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return 'NO'
    return "YES" if not stack else"NO"
print(is_valid_brackets("()))()[]"))

```

```python
#W7A28
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0: return False
    return True

def longest_composite_sequence(N):
    max_len = 0
    result = []
    for start in range(N+1, N+100):  # duyệt từ N+1 đến N+100
        temp = []
        for i in range(start, start+10):  # kiểm tra 10 số liên tiếp
            if not is_prime(i):
                temp.append(i)
            else:
                break
        if len(temp) > max_len:
            max_len = len(temp)
            result = temp
    return result

# Ví dụ
print(longest_composite_sequence(50))  # [24, 25, 26, 27, 28]
```
