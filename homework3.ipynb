```python
#bài14
# Initialize max and min values. Using the first input is a reliable method.
# A flag 'first_input' is used to handle the very first number.
first_input = True
max_val = None
min_val = None

print("Nhập vào các số nguyên (nhập -1 để dừng):")

while True:
    try:
        num = int(input())

        if num == -1:
            break

        if first_input:
            max_val = num
            min_val = num
            first_input = False
        else:
            if num > max_val:
                max_val = num
            if num < min_val:
                min_val = num
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")

# Print results after the loop terminates
if max_val is not None and min_val is not None:
    print(f"Số lớn nhất trong các số vừa nhập là: **{max_val}**")
    print(f"Số nhỏ nhất trong các số vừa nhập là: **{min_val}**")
else:
    print("Không có số hợp lệ nào được nhập.")
```

```python
#bài 17 cách 2
n = int(input("Nhập số nguyên dương n: "))
total_sum = 0
temp_n = n

while temp_n > 0:
    digit = temp_n % 10
    total_sum += digit
    temp_n //= 10

print(f"Tổng các chữ số của {n} là: **{total_sum}**")
```

```python
#bài19
def find_max(a):
    fib1, fib2 = 1, 1
    while fib2 <= a:
        fib1, fib2 = fib2, fib1 + fib2
    return fib1
a = int(input())
print(find_max(a))
```

```python
#bài20 
def kt(n):
    tong = 0
    for digit_char in n:
        digit = int(digit_char)
        if digit % 2 != 0 and digit != 1 and digit != 9:
            tong += 1
    return(tong)
n = input()
print(kt(n))

```
