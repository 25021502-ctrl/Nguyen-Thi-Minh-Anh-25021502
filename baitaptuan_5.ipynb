```python
#W5A1
def solon(a, b):
    return(max(a,b))
a, b = map(int,input().split())
print(solon(a, b))


```

```python
#W5A2
def swap(a,b):
    a, b = b, a
    return (a,b)
a, b = map(int,input().split())
print(swap(a,b))
```

```python
#W5A3
def kt(a):
    if a < 2:
        print('nhập số nguyên dương lớn hơn 2')
    else:
        for i in range (2, a):
            if a % i == 0:
                return False
        return True

a = int(input())
print(kt(a))
        
```

```python
#W5A4
def so_hoan_hao(n):
    tong = 0 
    for i in range(1,n):
        if n % i == 0:
            tong += i
    if tong == n:
        return True
    else:
        return False
n = int(input())
print(so_hoan_hao(n))
```

```python
#W5A5
def tim(lst,k):
    for i in range (len(lst)):
        if lst[i] == k:
            return i
    return -1
lst = list(map(int,input().split()))
k = int(input())
print(tim(lst,k))

```

```python
#W5A6
def giai_thua(n):
    tich = 1
    for i in range(1,n+1):
        tich *= i
    return tich
n = int(input())
print(giai_thua(n))
```

```python
#W5A7
def gia_lap(num1,num2,operat):
    if operat == '+':
        result = num1 + num2
    elif operat == '-':
        result = num1 - num2
    elif operat == '*':
        result = num1 * num2
    elif operat == ':':
        if num2 == 0:
            return('lỗi')
        result = num1/num2
    return round(result,2)
num1 = int(input())
num2 = int(input())
operat = input()
print(gia_lap(num1,num2,operat))
    
```

```python
#W5A9
def ting(n):
    tong = 0
    for digit_char in n:
        digit = int(digit_char)
        tong += digit
    return tong
n = input()
print(ting(n))

   
```
