```python
#W6A9
dict1 = eval(input("nhập dictionary thứ nhất(ví dụ: {'a':1})"))
dict2 = eval(input("nhập dictionary thứ 2(ví dụ: {'b': 3,})"))
merged_dict = {}
for key in dict1:
    merged_dict[key] = dict1[key]
for key in dict2:
    if key in merged_dict:
        merged_dict[key] += dict2[key]
    else:
        merged_dict[key] = dict2[key]
sorted_dict = dict(sorted(merged_dict.items()))
print(sorted_dict)
```

```python
#W6A10 lý thuyết: nhiệt độ và cân bằng nhiệt 
# biến đổi pha,cơ chế truyền nhiệt, tính chất phân tử của vâkj chất, mô hình động học phân tử khí lí tưởng, nhiệt dung,các pha của vật chất,cácloai quá trình nhiệt độ
def merge_dicts(dict1, dict2):
    merged = {}
    # Gộp các key từ cả hai dict
    for key in set(dict1.keys()).union(dict2.keys()):
        merged[key] = dict1.get(key, 0) + dict2.get(key, 0)
    # Sắp xếp theo thứ tự từ điển tăng dần
    sorted_merged = dict(sorted(merged.items()))
    return sorted_merged

# Ví dụ nhập liệu
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'b': 3, 'c': 4, 'd': 5}

result = merge_dicts(dict1, dict2)
print(result)
```

```python
#W6A11  
def split_even_odd(t):
    if not isinstance(t, tuple):
        raise TypeError("Đầu vào phải là tuple các số nguyên")
    evens = tuple(x for x in t if isinstance(x, int) and x % 2 == 0)
    odds = tuple(x for x in t if isinstance(x, int) and x % 2 != 0)
    return evens, odds
raw = input().strip()
data = tuple(int(s) for s in raw.split())
evens, odds = split_even_odd(data)
print(evens)
print(odds)
```

```python
#W6A12
from collections import Counter
def most_frequent_min_tiebreak(lst):
    if not isinstance(lst, list):
        raise TypeError ("Đầu vào phải là list các số nguyên.")
    if not lst:
        raise ValueError("list rỗng, không có phần tử để thống kê")
    freq = Counter()
    for x in lst:
        if not isinstance(x, int):
            raise TypeError("List phải chỉ chứa số nguyên")
        freq[x] +=  1
    max_count = max(freq.value())
    candidates = [num for num, c in freq.items() if c == max_count]
    return min(candidates)
raw = input().strip()
data = [int(s) for s in raw.split()]
print(most_frequent_min_tiebreak(data))
```

```python
#W6A13
# Nhập dictionary từ người dùng
original_dict = {}
n = int(input("Nhập số lượng phần tử trong dictionary: "))
for i in range(n):
    key = input(f"Nhập key thứ {i+1}: ")
    value = input(f"Nhập value thứ {i+1} (phải là duy nhất): ")
    original_dict[key] = value

# Đảo ngược key và value
reversed_dict = {value: key for key, value in original_dict.items()}

# In ra dictionary mới
print("Dictionary sau khi đảo ngược:")
print(reversed_dict)
```

```python
#W6A14
# Nhập hai dòng danh sách số nguyên từ người dùng
list1 = list(map(int, input("Nhập danh sách thứ nhất (cách nhau bởi dấu cách): ").split()))
list2 = list(map(int, input("Nhập danh sách thứ hai (cách nhau bởi dấu cách): ").split()))

# Tìm giao của hai danh sách, loại bỏ trùng lặp bằng cách dùng set
intersection = list(set(list1) & set(list2))

# In kết quả
print("Danh sách giao (không trùng lặp):", intersection)
```
