```python
#W4A6
import math
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(math.sqrt(n))+1):
        if n % i == 0:
            return False
    return True
a, b = map(int,input().split())
if a > b:
    a,b = b, a
tong = 0
for i in range (a,b+1):
    if is_prime(i):
        tong += i
print(tong)

```

```python
#W4A7
import math
def is_prime(k):
    if k <= 1:
        return False
    for i in range(2, int(math.sqrt(k)) +1):
        if k % i == 0:
            return False
    return True
n = int(input())
if n < 2:
    print('nhập n >= 2')
else:
    for divisor in range(n,1,-1):
        if n % divisor == 0 and is_prime(divisor):
            print(divisor)
            break
    
```

```python
#W4A8
# Nhập số nguyên dương n
n = int(input("Nhập số nguyên dương n: "))

steps = 0
current_n = n

# Vòng lặp cho đến khi số trở thành palindrome
while str(current_n) != str(current_n)[::-1]:
    # Đảo ngược số hiện tại bằng cách chuyển sang chuỗi, đảo ngược, rồi chuyển lại thành số
    reversed_n = int(str(current_n)[::-1])
    
    # Cộng số hiện tại với số đảo ngược
    current_n += reversed_n
    
    # Tăng số bước
    steps += 1
    
print(f"Số bước lặp: {steps}")
print(f"Giá trị palindrome thu được: {current_n}")
```

```python
#W4A9
def solve_w4a9(n):
    """
    Tìm và in ra các số chính phương nhỏ hơn hoặc bằng n 
    với tất cả các chữ số đôi một khác nhau.
    """
    results = []
    i = 1
    while True:
        # Sửa lỗi cú pháp gán giá trị
        square = i * i 
        if square > n:
            break
        
        s_square = str(square)
        # Kiểm tra tính duy nhất của chữ số bằng cách so sánh độ dài chuỗi và set
        if len(s_square) == len(set(s_square)):
            results.append(str(square))
            
        i += 1
    
    if not results:
        # Sửa lỗi in ấn "no"
        print("(no number)")
    else:
        # Loại bỏ dấu chấm thừa ở cuối
        print(" ".join(results))
n = int(input("Nhập số nguyên dương n: "))
print(solve_w4a9(n)) 
# solve_w4a9(100)
# solve_w4a9(200)
```

```python
#W4A10
def collatz_length(num):
    """Tính độ dài dãy Collatz cho một số."""
    length = 1
    while num != 1:
        if num % 2 == 0:
            num //= 2
        else:
            num = 3 * num + 1
        length += 1
    return length

def find_longest_collatz(n):
    """Tìm số có độ dài dãy Collatz dài nhất trong khoảng [1, n]."""
    max_length = 0
    best_x = 0
    # Bắt đầu từ 1 đến n (bao gồm n)
    for x in range(1, n + 1):
        current_length = collatz_length(x)
        if current_length > max_length:
            max_length = current_length
            best_x = x
        elif current_length == max_length:
            # Nếu bằng độ dài, chọn số nhỏ nhất
            if best_x == 0 or x < best_x:
                best_x = x
    return best_x, max_length

# Nhập n từ người dùng
n_input = int(input("Nhập vào số nguyên dương n: "))
x, L = find_longest_collatz(n_input)
print(f"{x} {L}")
```
